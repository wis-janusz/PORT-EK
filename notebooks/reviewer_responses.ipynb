```python
deer_fasta = SeqIO.parse("../input/deer/EPI_SET_240422va.fasta", "fasta")
Ns = []
for record in deer_fasta:
    Ns.append(record.seq.count("N"))

fig, ax = plt.subplots()
sns.histplot(Ns)
ax.set_xlabel("Number of ambigous NTs")
ax.set_ylabel("Number of sequences")
plt.savefig("../output/deer/deer_Ns.png")
print(min(Ns))
print(np.median(Ns))
print(np.mean(Ns))
print(np.max(Ns))
print(len([N for N in Ns if N > 0]))
```
